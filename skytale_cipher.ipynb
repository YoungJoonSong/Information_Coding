```python
import math

def encrypt_scytale(plaintext, key):
    # 공백을 제거하거나 유지할 수 있지만, 여기서는 이해를 돕기 위해 유지합니다.
    # 막대의 굵기(key)만큼의 열을 가진 행렬로 생각합니다.
    n = len(plaintext)
    columns = key
    rows = math.ceil(n / columns)
    
    # 부족한 칸은 패딩('*')으로 채웁니다.
    padded_text = plaintext.ljust(rows * columns, '*')
    
    ciphertext = ""
    # 열 방향으로 읽어서 암호문을 생성합니다.
    for col in range(columns):
        for row in range(rows):
            idx = row * columns + col
            ciphertext += padded_text[idx]
            
    return ciphertext

def decrypt_scytale(ciphertext, key):
    n = len(ciphertext)
    columns = key
    rows = n // columns
    
    plaintext = ""
    # 암호화의 역순으로, 행 방향으로 다시 재배치합니다.
    for row in range(rows):
        for col in range(columns):
            idx = col * rows + row
            plaintext += ciphertext[idx]
            
    return plaintext.rstrip('*') # 패딩 제거

# 테스트 실행
message = "SENDHELPQUICKLY" # 평문
key_value = 4                # 막대 굵기 (Key)

encrypted = encrypt_scytale(message, key_value)
decrypted = decrypt_scytale(encrypted, key_value)

print(f"평문: {message}")
print(f"암호문: {encrypted}")
print(f"복호화 결과: {decrypted}")
```

```python
import math

class ScytaleCipher:
    def __init__(self, key):
        """
        막대의 굵기(Key)를 설정하여 암호화 객체를 생성합니다.
        :param key: 막대의 굵기 (정수)
        """
        self.key = int(key)

    def encrypt(self, plaintext):
        """
        평문을 스키테일 방식으로 암호화합니다.
        """
        # 공백을 포함한 전체 길이 계산
        n = len(plaintext)
        columns = self.key
        rows = math.ceil(n / columns)
        
        # 행렬을 채우기 위해 부족한 부분은 패딩('*') 처리
        padded_text = plaintext.ljust(rows * columns, '*')
        
        ciphertext = ""
        # [실습 포인트] 행렬의 열(Column)을 기준으로 순회하며 글자를 추출합니다.
        for col in range(columns):
            for row in range(rows):
                index = row * columns + col
                ciphertext += padded_text[index]
        
        return ciphertext

    def decrypt(self, ciphertext):
        """
        암호문을 동일한 굵기의 막대로 복호화합니다.
        """
        n = len(ciphertext)
        columns = self.key
        rows = n // columns
        
        plaintext = ""
        # [실습 포인트] 암호화된 문자열을 다시 행(Row) 단위로 재배치합니다.
        for row in range(rows):
            for col in range(columns):
                index = col * rows + row
                plaintext += ciphertext[index]
        
        # 우측의 패딩('*') 제거 후 반환
        return plaintext.rstrip('*')

# --- 실습 실행부 ---
if __name__ == "__main__":
    print("=== [정보보호 실습] 스키테일 암호 알고리즘 ===")
    
    try:
        # 1. 사용자로부터 메시지와 키 입력 받기
        user_input = input("암호화할 메시지를 입력하세요: ")
        user_key = int(input("막대의 굵기(Key)를 입력하세요 (숫자): "))
        
        # 2. 암호화 객체 생성
        scytale = ScytaleCipher(user_key)
        
        # 3. 암호화 수행
        encrypted = scytale.encrypt(user_input)
        print(f"\n[암호화 단계] 가죽 띠에 적힌 문자: {encrypted}")
        
        # 4. 복호화 수행
        decrypted = scytale.decrypt(encrypted)
        print(f"[복호화 단계] 복구된 메시지: {decrypted}")
        
        # 5. 검증
        if user_input == decrypted:
            print("\n결과: 암호화 및 복호화가 성공적으로 완료되었습니다.")
        else:
            print("\n결과: 복호화된 메시지가 원문과 다릅니다.")
            
    except ValueError:
        print("오류: 막대 굵기는 반드시 숫자로 입력해야 합니다.")
```
