# 상호정보량

정보이론에서 **상호정보량(Mutual Information, $I(X; Y)$)** 은 두 확률 변수 사이의 상호 의존성을 측정하는 척도입니다. 쉽게 말해, 한 변수를 알게 되었을 때 다른 변수에 대해 얼마나 많은 정보를 얻을 수 있는지를 수치화한 것입니다.

---

## 1. 핵심 개념
상호정보량은 두 변수가 얼마나 결합되어 있는지를 나타냅니다.
*   만약 두 변수가 완전히 **독립적**이라면, 상호정보량은 **0**이 됩니다. (한 변수가 다른 변수에 대해 아무런 정보도 주지 않음)
*   두 변수가 밀접하게 연관될수록 상호정보량 값은 커집니다.

## 2. 수식적 정의
상호정보량은 엔트로피($H$)를 사용하여 여러 방식으로 정의할 수 있습니다.

### 엔트로피와의 관계
가장 직관적인 정의는 다음과 같습니다:
$$I(X; Y) = H(X) - H(X|Y)$$

*   **$H(X)$**: $X$의 불확실성(엔트로피)
*   **$H(X|Y)$**: $Y$를 알 때 $X$의 남아있는 불확실성(조건부 엔트로피)

즉, $X$의 전체 불확실성에서 $Y$를 알고 난 후의 불확실성을 뺀 값이 바로 '공유하는 정보량'이 됩니다.

### 대칭성 (Symmetry)
상호정보량은 대칭적입니다. $X$가 $Y$에 대해 주는 정보량은 $Y$가 $X$에 대해 주는 정보량과 같습니다.
$$I(X; Y) = I(Y; X) = H(X) + H(Y) - H(X, Y)$$
 여기서 $H(X, Y)$는 두 변수의 **결합 엔트로피(Joint Entropy)** 입니다.

---

## 3. 벤 다이어그램을 통한 이해
상호정보량은 흔히 두 원이 겹쳐진 벤 다이어그램으로 시각화합니다.



*   **왼쪽 원**: $H(X)$
*   **오른쪽 원**: $H(Y)$
*   **두 원의 전체 합집합**: 결합 엔트로피 $H(X, Y)$
*   **교집합 부분**: **상호정보량 $I(X; Y)$**

---

## 4. 주요 특징 및 활용
*   **비음수성 ($I(X; Y) \ge 0$):** 정보량은 결코 음수가 될 수 없습니다. 최소 0(독립일 때)입니다.
*   **데이터 과학 및 머신러닝에서의 활용:**
    *   **특성 선택(Feature Selection):** 타겟 변수와 상호정보량이 높은 변수를 선택하여 모델의 성능을 높입니다.
    *   **독립성 검정:** 두 데이터 군 사이의 상관관계를 비선형적인 관계까지 포함하여 파악할 때 유용합니다.
*   **통신 이론:** 채널 용량(Channel Capacity)을 정의할 때 핵심적인 역할을 합니다. 송신 신호 $X$와 수신 신호 $Y$ 사이의 상호정보량을 최대화하는 것이 통신의 목적입니다.

상호정보량은 상관계수(Correlation)와 비슷해 보이지만, 상관계수는 선형적인 관계만 측정하는 반면 상호정보량은 **모든 종류의 통계적 의존성**을 포착한다는 점에서 훨씬 강력한 도구입니다.

# 결합 엔트로피

정보이론에서 **결합 엔트로피(Joint Entropy, $H(X, Y)$)** 는 두 개 이상의 확률 변수가 결합된 시스템 전체의 불확실성을 측정하는 척도입니다.

단일 변수의 엔트로피가 하나의 사건이 가진 정보량을 나타낸다면, 결합 엔트로피는 **"두 사건이 동시에 일어날 때의 전체 정보량"** 이 얼마인지를 말해줍니다.

---

## 1. 수식적 정의
두 이산 확률 변수 $X$와 $Y$의 결합 확률 분포를 $p(x, y)$라고 할 때, 결합 엔트로피는 다음과 같이 정의됩니다.

$$H(X, Y) = -\sum_{x \in \mathcal{X}} \sum_{y \in \mathcal{Y}} p(x, y) \log p(x, y)$$

이 수식은 모든 가능한 $(x, y)$ 조합에 대해 그 발생 확률과 정보량($-\log p(x, y)$)의 기댓값을 구한 것입니다.

---

## 2. 주요 성질과 관계식
결합 엔트로피는 개별 엔트로피($H(X), H(Y)$), 조건부 엔트로피($H(Y|X)$), 그리고 상호정보량($I(X; Y)$)과 밀접한 관계를 가집니다.

### ① 연쇄 법칙 (Chain Rule)
시스템 전체의 불확실성은 한 변수의 불확실성에 그 변수가 주어졌을 때 나머지 변수의 불확실성을 더한 것과 같습니다.
$$H(X, Y) = H(X) + H(Y|X)$$
$$H(X, Y) = H(Y) + H(X|Y)$$

### ② 독립성과의 관계
두 변수가 서로 **독립(Independent)** 이라면, 전체 불확실성은 각 변수의 불확실성을 단순히 더한 것과 같습니다.
$$H(X, Y) = H(X) + H(Y)$$
반대로 두 변수가 의존적이라면, 중복되는 정보(상호정보량)만큼 전체 엔트로피는 줄어듭니다.
$$H(X, Y) = H(X) + H(Y) - I(X; Y)$$



---

## 3. 직관적인 이해
*   **완전 독립일 때:** $X$가 '주사위 결과'이고 $Y$가 '내일 날씨'라면, 두 사건은 아무 상관이 없습니다. 이때 결합 엔트로피는 두 정보를 각각 알기 위해 필요한 노력의 합과 같습니다.
*   **완전 종속일 때:** 만약 $Y$가 항상 $X$와 같은 값을 가진다면($Y=X$), $X$를 아는 순간 $Y$도 알게 되므로 결합 엔트로피 $H(X, Y)$는 단일 엔트로피 $H(X)$와 같아집니다.
*   **상한선:** 결합 엔트로피는 항상 각 개별 엔트로피의 합보다 작거나 같습니다 ($H(X, Y) \leq H(X) + H(Y)$).

## 4. 활용 사례
*   **데이터 압축:** 두 변수 사이의 상관관계가 높을수록 결합 엔트로피가 낮아지며, 이는 데이터를 더 효율적으로 압축할 수 있음을 의미합니다.
*   **통신 시스템:** 송신 신호와 수신 신호 사이의 노이즈를 포함한 전체 시스템의 복잡도를 분석할 때 사용됩니다.
*   **머신러닝:** 다변수 확률 분포를 다루는 모델에서 변수 간의 총체적인 불확실성을 평가하는 지표로 활용됩니다.

# 조건부 엔트로피

정보이론에서 **조건부 엔트로피(Conditional Entropy, $H(Y|X)$)** 는 확률 변수 $X$의 값을 알고 있을 때, 또 다른 확률 변수 $Y$가 가질 수 있는 평균적인 불확실성을 의미합니다.

쉽게 말해, **"어떤 정보($X$)가 주어진 상태에서, 여전히 모르는 정보($Y$)는 얼마나 되는가?"** 를 측정하는 척도입니다.

---

## 1. 수식적 정의
조건부 엔트로피 $H(Y|X)$는 $X$의 각 결과값 $x$에 대하여 $Y$의 엔트로피를 계산한 후, 이를 $X$가 발생할 확률로 가중 평균하여 구합니다.

$$H(Y|X) = \sum_{x \in \mathcal{X}} p(x) H(Y|X=x)$$
$$H(Y|X) = -\sum_{x \in \mathcal{X}} \sum_{y \in \mathcal{Y}} p(x, y) \log p(y|x)$$

이 값은 $X$를 통해 $Y$를 예측하려 할 때, 여전히 남아있는 **'노이즈'** 나 **'모호함'** 의 양으로 해석할 수 있습니다.

---

## 2. 다른 엔트로피와의 관계
조건부 엔트로피는 결합 엔트로피 및 단일 엔트로피와 매우 긴밀한 관계를 맺고 있습니다.

### ① 결합 엔트로피와의 관계 (Chain Rule)
전체 시스템의 불확실성($H(X, Y)$)에서 $X$에 대한 불확실성을 빼고 남은 것이 바로 $Y$의 조건부 엔트로피입니다.
$$H(Y|X) = H(X, Y) - H(X)$$

### ② 상호정보량과의 관계
$Y$가 원래 가지고 있던 전체 불확실성($H(Y)$)에서 $X$를 통해 알게 된 정보량($I(X; Y)$)을 빼면 조건부 엔트로피가 됩니다.
$$H(Y|X) = H(Y) - I(X; Y)$$



---

## 3. 직관적인 예시
*   **완전한 예측 가능 ($H(Y|X) = 0$):**
    만약 $X$가 '비가 온다'는 사실이고, $Y$가 '지면이 젖는다'는 사실이라면, $X$를 아는 순간 $Y$는 확실해집니다. 이때 불확실성은 0이 됩니다.
*   **아무런 도움이 안 됨 ($H(Y|X) = H(Y)$):**
    만약 $X$가 '오늘 아침 식사 메뉴'이고, $Y$가 '주식 시장의 등락'이라면, $X$를 안다고 해서 $Y$에 대한 불확실성이 줄어들지 않습니다. 즉, 두 변수가 **독립**일 때 발생합니다.
*   **일반적인 경우:**
    $X$가 '구름의 양'이고 $Y$가 '강수 여부'라면, 구름의 양을 아는 것이 강수 여부를 예측하는 데 도움은 되지만 100% 확실하지는 않습니다. 이때 조건부 엔트로피는 $H(Y)$보다는 작지만 0보다는 큰 값을 가집니다.

---

## 4. 특징 및 성질
*   **정보는 불확실성을 낮춘다 (Information Can't Hurt):**
    어떤 정보 $X$를 아는 것은 $Y$에 대한 불확실성을 줄이거나 최소한 유지할 뿐, 결코 늘리지 않습니다.
    $$H(Y|X) \leq H(Y)$$
*   **비음수성:** 엔트로피와 마찬가지로 항상 0 이상의 값을 가집니다.

## 5. 활용 분야
*   **머신러닝 (Decision Tree):** 의사결정나무 모델에서 어떤 특성(Feature)으로 데이터를 나눌지 결정할 때, 조건부 엔트로피가 가장 낮아지는(즉, 정보 이득이 가장 큰) 방향을 선택합니다.
*   **통신 공학:** 채널을 통해 신호를 보낼 때 발생하는 노이즈의 양을 측정하는 데 사용됩니다. 송신 신호가 주어졌을 때 수신 신호의 조건부 엔트로피가 낮을수록 정확한 통신이 가능합니다.
